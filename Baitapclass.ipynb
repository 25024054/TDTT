Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class MyDate:
    def __init__(self,dd,mm,yy):
        self.d=dd
        self.m=mm
        self.y=yy
    def checkValid(self):
        maxday=-1
        match self.m :
         case 1|3|5|7|8|10|12:
            maxday=31
         case 4|6|9|11:
            maxday=30
         case 2:
          if (self.y%4==0 and self.y%100!=0) or (self.y%400==0):
            maxday=29
          else:
            maxday=28
        return 1 <= self.d <= maxday and self.y >0
    def getNextday(self):
        if (self.checkValid()):
            day1=MyDate(self.d+1,self.m,self.y)
            if(day1.checkValid()):
             return day1
            day1=MyDate(1,self.m+1,self.y)
            if(day1.checkValid()):
                return day1
            return MyDate(1,1,self.y)
    def __str__(self):
        return f"{self.d:02d}/{self.m:02d}/{self.y:04d}"
d,m,y=map(int,input().split("/"))
day1=MyDate(d,m,y)
print(day1.checkValid())
day2=day1.getNextday()
print(day2)
```

Input 
w h k (số thực dương)

```python
class Rectangle():
    def __init__(self,w,h):
        self.w=w
        self.h=h
    def area(self):
        return self.w*self.h
    def perimeter(self):
        return 2*(self.w+self.h)
    def scale(self,k):
        self.w=self.w*k
        self.h=self.h*k
w,h,k=map(float,input().split())
r1=Rectangle(w,h)
r1.scale(k)
print(f"Dien tich sau scale la : {r1.area():.2f}")
print(f"Chi vi sau scale la : {r1.perimeter():.2f}")
```

### W8A7 Yêu cầu Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:  
Rút gọn phân số trong __init__ (quy ước mẫu > 0).

Các phương thức add/sub/mul/div trả về phân số đã rút gọn. 

__str__() in dạng a/b. 

Input 

a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/}) 
 
Output 
Phân số kết quả đã rút gọn 

```python
import math
class Fraction:
    def __init__(self, num, den):
        if den < 0:
            num = -num
            den = -den
        common = math.gcd(abs(num), abs(den))       
        self.num = num // common
        self.den = den // common
    def add(self, other):
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def sub(self, other):
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def mul(self, other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def div(self, other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)
    def __str__(self):
        return f"{self.num}/{self.den}"
    # hàm cuối là để định dạng các hàm trên nếu không sẽ phải print result.num và result.den
data = input().split()
a = int(data[0])
b = int(data[1])
op = data[2]        
c = int(data[3])
d = int(data[4])
f1 = Fraction(a, b)
f2 = Fraction(c, d)
if op == '+':
    result = f1.add(f2)
elif op == '-':
    result = f1.sub(f2)
elif op == '*':
    result = f1.mul(f2)
elif op == '/':
    result = f1.div(f2)
print(result)
```

Input 
Dòng 1: name 
Dòng 2: n 
n dòng tiếp: subject score (score là số thực 0–10) 
Output 
name avg rank (avg 2 chữ số). 

```python
class Student :
    def __init__(self,name):
        self.name=name
        self.scores={}
    def add_score(self,subject,score):
        self.scores[subject]=float(score)
    def avg(self):
        if not self.scores:
            return 0.0
        tong_diem = sum(self.scores.values())
        so_luong_mon = len(self.scores)
        return tong_diem/so_luong_mon
    def rank(self):
        if self.avg() >= 8:
            return "Excellent"
        elif self.avg() >= 6.5: 
            return "Good"
        elif self.avg() >= 5:   
            return "Average"
        else:
            return "Poor"
name=input().strip()
s1=Student(name)
n=int(input())
for i in range(n):
    s=input().split()
    subject=s[0]
    score=s[1]
    s1.add_score(subject,score)
print(f"{s1.name} {s1.avg():.2f} {s1.rank()}")


```

### W8A12 [Complex] Khai báo kiểu dữ liệu biểu diễn số phức \(ComplexNumber\) bao gồm \(2\) biến \(realPart\) và \(imaginaryPart\) kiểu \(double\) biểu diễn phần thực và phần ảo của một số phức. Viết hàm khởi tạo ComplexNumber(double initRealPart, double initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo của một số phức và hàm ComplexNumber multiply(complexNumber other) trả về tích của số phức hiện tại và một số phức khác.  
Đầu vào 

Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực 
cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo 
của một số phức. Các số trong đầu vào có giá trị tuyệt đối 
không vượt quá \(100\). 

Đầu ra 

In ra màn hình hai số biểu diễn phần thực và phần ảo của tích 
hai số phức

```python
class ComplexNumber():
    def __init__(self,initRealPart, initImaginaryPart):
        self.realPart = initRealPart
        self.imaginaryPart = initImaginaryPart
    def multiply(self,other):
        newReal = (self.realPart * other.realPart) - (self.imaginaryPart * other.imaginaryPart)
        newImaginary = (self.realPart * other.imaginaryPart) + (self.imaginaryPart * other.realPart)
        return ComplexNumber(newReal, newImaginary)
    # khong thể trả vì return vì return chỉ trả về đc 1 trong 2
r1, i1 = map(float, input().split())
z1 = ComplexNumber(r1, i1)
r2, i2 = map(float, input().split())
z2 = ComplexNumber(r2, i2)
result = z1.multiply(z2)
def format_num(n):
    if n % 1 == 0:     
        return int(n) 
    else:
        return n
print(f"{format_num(result.realPart)} {format_num(result.imaginaryPart)}")
```

### W8A13 [Triangle] 
Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa 3 biến 
\(egde1,\,edge2,\,edge3\) kiểu \(int\) là độ dài \(3\) cạnh của 
tam giác đó. 
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int 
initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm 
double getArea() trả về diện tích của tam giác đó (nếu tam giác 
không hợp lệ, trả về \(-1\)). 
Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một 
tam giác và in ra diện tích của tam giác đó. 
Đầu vào 
Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài 
\(3\) cạnh của tam giác 
Đầu ra 
In ra màn hình diện tích tam giác (làm tròn đến\(2\) chữ số thập 
phân sau dấu phẩy). In ra \(invalid\) nếu \(3\) số nhập vào 
không phải là \(3\) cạnh của một tam giác 

```python
class Triangle():
    def __init__(self,edge1,edge2,edge3):
        self.edge1=float(edge1)
        self.edge2=float(edge2)
        self.edge3=float(edge3)
    def getArea(self):
        if self.edge1+self.edge2<=self.edge3 or self.edge1+self.edge3<=self.edge2 or self.edge3+self.edge2<=self.edge1:
            return -1
        else:
            p=(self.edge1+self.edge2+self.edge3)/2
            s=(p*(p-self.edge1)*(p-self.edge2)*(p-self.edge3))**0.5
            return s
edge1,edge2,edge3=map(int,input().split())
triangle=Triangle(edge1,edge2,edge3)
if triangle.getArea()==-1:
  print("invalid")
else:
    print(f"{triangle.getArea():.2f}")
```
