### W10A3 [ValueError - HappyNumbers] Nhập vào một số nguyên dương a, Kiểm tra xem a có phải là số hạnh phúc không. Một số nguyên dương n được gọi là số hạnh phúc nếu ta thực hiện liên tiếp các bước sau:
- Tính tổng bình phương các chữ số của n.
- Lặp lại bước đó với kết quả vừa tính.
- Nếu cuối cùng kết quả là 1, thì n là số hạnh phúc.
- Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1 (ví dụ như 4 → 16 → 37 → ...), thì n
không phải là số hạnh phúc.
Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback để gỡ lỗi nhập liệu. (Dùng try ... except
bắt lỗi ValueError thêm set) 


```python
try:
    n=int(input())
    checkvohan=set()
    while n!=1 and n not in checkvohan:
      checkvohan.add(n)
      tong=0
      for j in str(n):
         tong+=int(j)**2
      n=tong
    if n==1:
       print("la so hanh phuc")
    else:
        print("Khong phai so hanh phuc")
except ValueError:
   print("Nhập đúng số")


```

### W10A4 Sử dụng try except,kết hợp cùng else... finally. Tính tổng 2 số, nếu nhập sai kiểu thì thông báo, luôn luôn in "Kết thúc chương trình." sau khi chạy

```python
try:
    a,b=map(int,input().split())
except ValueError:
    print("Vui lòng nhập chữ số")
else:
    print(a+b)
finally:
    print("Kết thúc chương trình")
```

### W10A10 Viết chương trình nhập vào phần tử N (N > 0) và một mảng gồm N số nguyên dương. Một mảng là hợp lệ nếu nó gồm các số khác nhau. In ra tổng các phần tử nếu mảng đó là hợp lệ. Nếu mảng không hợp lệ, in ra "Mang khong hop le".


```python
n=int(input())
a=[]
x=[]
for i in range(n):
    b=int(input())
    a.append(b)
for j in a:
    if j not in x:
        x.append(j)
if len(x)==len(a):
    print(sum(x))
else:
    print("Mang khong hop le")
```

### W10A15 [Raise_Exception - PositiveNumber] Nhập vào một số nguyên n. Nếu n <= 0 thì chủ động phát sinh ngoại lệ và in ra "So phai duong". Nếu hợp lệ thì in ra bình phương của n.
