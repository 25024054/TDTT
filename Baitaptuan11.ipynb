### W11A1"""Bất cứ một số nguyên dương nào cũng tồn tại một số nguyên dương là số đảo ngược của nó. Hãy viết hàm def reverse (path: str) -> int: để tìm số nguyên đảo ngược của số n .Đầu vào của hàm là một xâu biểu diễn đường dẫn đến tệp. Nội dung tệp chứa duy nhất một số nguyên dương.Đầu ra : Hàm trả về số nguyên dương đảo ngược của số trong tệp đầu vào."""

```python
def reverse(path: str) -> int:
    with open(path, 'r') as file:
        number_str = file.read().strip()
    reversed_str = number_str[::-1]
    return int(reversed_str)
```

### W11A4 Hai chuỗi là anagram của nhau nếu các chữ cái của chuỗi này có thể được sắp xếp để tạo thành chuỗi kia. Hãy viết hàm def areAnagrams(path1: str, path2: str) trả về true nếu hai chuỗi đầu vào là anagram của nhau ngược lại false. Hàm không phân biệt chữ hoa chữ thường và không quan tâm đến các dấu câu và dấu cách.
Đầu vào là 2 tên file trên 2 dòng, mỗi file chứa một chuỗi kí tự.
Đầu ra: là True nếu 2 chuỗi kí tự là anagram của nhau, False nếu ngược lại

```python
def areAnagrams(path1: str, path2: str) -> bool:
    with open(path1, 'r') as f1:
        str1 = f1.read()
    with open(path2, 'r',) as f2:
        str2 = f2.read()
    clean_str1 = ""
    for char in str1:
        if char.isalpha():              
            clean_str1 += char.lower()            
    clean_str2 = ""
    for char in str2:
        if char.isalpha():
            clean_str2 += char.lower()
    if sorted(clean_str1) == sorted(clean_str2):
        return True
    return False
```

### W11A5 Viết hàm def count(path: str, character: str), nhận tham số truyền vào là một tên file và một kí tự character, trả về số lần xuất hiện của kí tự character trong chuỗi sentence đọc được từ file

```python
def count(path: str, character: str):
    with open(path, 'r') as file:
        f1=file.read().strip()
    count_=0
    for i in f1:
        if i==character:
            count_+=1
    return count_
        
```

### W11A6 Việc đầu tiên bạn cần làm là viết một hàm def countCharacter(path: str, letter: str) nhận tham số đầu vào là đường dẫn đến tệp văn bản và kí tự cần đếm số lần xuất hiện, hàm trả về số lần xuất hiện của kí tự truyền vào.Lưu ý: bạn cần đếm số lần xuất hiện của kí tự mà không phân biệt chữ hoa chữ thường.

```python
def countCharacter(path: str, letter: str):
    with open(path, 'r') as file:
        f1=file.read().strip()
    count_=0
    for i in f1.lower():
        if i==letter.lower():
            count_+=1
    return count_
```

### W11A8 Viết chương trình in các từ Fibonacci thứ 0 đến n và ghi ra file output.txt. Công thức của dãy Fibonacci như sau:
f(0)=a
f(1)=b
f(n)=f(n−1)+f(n−2)

```python
def generate_fibonacci_words(n: int):
    words = [] 
    if n >= 0:
        words.append("a")  # f(0) = a
    if n >= 1:
        words.append("b")  # f(1) = b     
    for i in range(2, n + 1):
        next_word = words[i-1] + words[i-2]
        words.append(next_word)
    with open("output.txt", "w", encoding="utf-8") as file:
        for i in range(len(words)):
            print(f"f({i}) = {words[i]}")
            file.write(words[i] + "\n")           
    print("\n=> Đã ghi thành công vào file output.txt!")
```

### W11A12 Với một dãy số nguyên cho trước, hãy viết hàm def maximumProduct (path: str) -> int: tìm bộ ba số có tích lớn nhất trong tất cả các bộ số và trả về tích lớn nhất đó. Hàm nhận đầu vào là đường dẫn đến tệp lưu mảng số nguyên cho trước. Hàm trả về tích lớn nhất có thể có. Dòng đầu tiên của tệp chứa kích cỡ n, dòng kế tiếp chứa n số nguyên cách nhau bởi một khoảng trắng. 


```python
def maximumProduct(path: str) -> int:
    with open(path, 'r', encoding='utf-8') as file:
        n = int(file.readline().strip())
        nums = [int(x) for x in file.readline().split()]       
    nums.sort()
    tich_3_so_cuoi = nums[-1] * nums[-2] * nums[-3]
    tich_2_so_dau_va_1_so_cuoi = nums[0] * nums[1] * nums[-1]
    return max(tich_3_so_cuoi, tich_2_so_dau_va_1_so_cuoi)
```

### W11A14 Kết quả được ghi trong một bảng biểu và được lưu trữ trong một tệp .csv. Cột đầu tiên của bảng là số thứ tự vòng đua (lap) và cột tiêp theo là thời gian Frank hoàn thành vòng đua đó, ví dụ một bảng và cấu trúc tệp tương ứng như sau:

Lap,Time

1,143.23

2,134.12

Cuối cùng, Frank cần phải tính thời gian trung bình anh hoàn thành vòng đua. Hãy giúp Frank bằng cách viết hàm def averageTime(path: str) , nhận tham số là xâu kí tự lưu đường dẫn tới tệp .csv và trả về thời gian trung bình hoàn thành vòng đua. Lưu ý, tùy tâm trạng Frank sẽ chạy thử xe nhiều vòng hay ít vòng, nên số vòng đua trong các tệp .csv là không cố định. Kết quả lấy đến 2 chữ số phần thập phân. 

```python
def averageTime(path: str):
    with open(path, 'r') as file:
        next(file)      
        tong_thoi_gian = 0.0
        so_vong = 0       
        for line in file:
            if not line.strip():
                continue          
            row = line.split(',')
            thoi_gian = float(row[1].strip())
            tong_thoi_gian += thoi_gian
            so_vong += 1
    trung_binh = tong_thoi_gian / so_vong
    return round(trung_binh, 2)
```

### W11A15 Hãy viết hàm def findMovies (path: str) -> list[str] giúp khán giả tìm ra phim với điểm đánhgiá lớn hơn 8.0 và có mô tả Không phải là "nhàm chán" ("boring").Hàm nhận đầu vào là đường dẫn đến tệp chứa các poster phim. Hàm trả lại các tênnhững phim thỏa mãn điều kiện trên. Nếu không tìm thấy phim nào phù hợp, trả về thôngbáo "All are boring!". Kết quả in theo thứ tự xuất hiện trước sau trong file đầu vào
Tệp có cấu trúc như sau:
id, movie, description, rating
1, Thor, great 3D, 8.0
2, The Rings, Fantasy, 7.5
3, Irish, boring, 6.2
4, Ice song, Fantasy, 8.6
5, Home Alone, Interesting, 6.5 

```python
def findMovies(path: str):
    with open(path, 'r') as file:
        next(file)
        a=[]
        for line in file:
            if not line.strip():
                continue    
            row = line.rstrip().split(",")
            if float(row[3].strip()) >8.0 and row[2].strip()!="boring":
                a.append(row[1].strip())
    if len(a)==0:
        return "All are boring!"
    else:
        return a
```
