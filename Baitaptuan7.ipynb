### W7A3 - Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection Sort) 

```python
def bubble_sort(arr):
    for i in range(len(arr)):
     for j in range(0, len(arr) - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]     
    return arr
arr=list(map(int,input().split()))
print(bubble_sort(arr))
# cứ sau mỗi i thì thằng to nhất đc đẩy ra sau cùng
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j 
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: binary search in the queries list 

```python
arr=eval(input())
q=list(map(int,input().split()))
a=[]
for i in q:
 b=[]
 for j in arr:
  if j[0]<=i and j[-1]>=i:
   b.append(j)
 if len(b)==0:
  a.append(-1)
 else:
  min=b[0][-1]-b[0][0]+1
  for k in b: 
   if k[-1]-k[0]+1<min:
    min=k[-1]-k[0]+1
  a.append(min)
print(a)
```

```python
def dem(arr):
 a=[]
 if not arr:  # bị dính nếu nhập mảng rỗng
  return []
 for i in range (len(arr)):
  count=0
  for j in range(i+1,len(arr)):
   if arr[i]>arr[j]:
    count=count+1
  a.append(count)
 return a
arr=list(map(int,input().split()))
print(dem(arr))
```
