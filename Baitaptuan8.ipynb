### W7A16 Cho một mảng số nguyên không âm. Gọimảng đó là "mảng đẹp" nếu mỗi phần tửx trong mảng xuất hiện đúng x lần.Kiểm tra xem mảng có phải là mảng đẹphay không. Nếu không hãy xóa đi y số (ylà nhỏ nhất có thể) để mảng còn lại thành mảng đẹp

```python
def beatiful_list(arr):
 a={}
 for i in arr:
  if i in a:
   a[i]+=1
  else:
   a[i]=1
 b=0
 for x,y in a.items():
  if y==x:
   continue
  elif y>x:
   b+=(y-x)
  elif y<x:
   b+=y
 if b==0:
  return "Yes"
 else:
  return (f"No {b}")
arr=list(map(int,input().split()))
print(beatiful_list(arr))
```

### W7A25 Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số 

```python
def find_max_k_digits(num_str, k):
    n = len(num_str)
    remove_count = n - k 
    stack = []
    for digit in num_str:
        while stack and remove_count > 0 and digit > stack[-1]:
            stack.pop() 
            remove_count -= 1 
        stack.append(digit)
    result_stack = stack[:k]
    return "".join(result_stack)
num_str=input().strip()
k=int(input())
print(find_max_k_digits(num_str, k))
```

### W7A27 Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không 

```python
def is_valid_parentheses(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '[', '>': '<'}
    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'
            if mapping[char] != top_element:
                return False
        else:
            stack.append(char)
    return len(stack)==0
s=input().strip()
if is_valid_parentheses(s):
    print("YES")
else:
    print("NO")
```

### W7A28 Bài toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua) 

```python
def is_composite(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return True
    return False
def longest_composite_sequence(N):
    max_seq = []       
    current_seq = []   
    for i in range(1, N):
        if is_composite(i):
            current_seq.append(i)
        else:
            if len(current_seq) > len(max_seq):
                max_seq = current_seq.copy()
            current_seq = []
    if len(current_seq) > len(max_seq):
        max_seq = current_seq
    return max_seq
n=int(input())
ket_qua = longest_composite_sequence(n)
print(*ket_qua,sep=', ')
```

### W7A29 Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau 

```python
def day_con_tang_lien_tuc(arr):
    if not arr:
        return []
    max_seq = []           
    current_seq = [arr[0]] 
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            current_seq.append(arr[i])
        else:
            if len(current_seq) > len(max_seq):
                max_seq = current_seq.copy()
            current_seq = [arr[i]]
    if len(current_seq) > len(max_seq):
        max_seq = current_seq
    return (max_seq)
arr=list(map(int,input().split()))
print(*day_con_tang_lien_tuc(arr))
```

### W7A30 Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím

```python
def day_con_tang_khong_lien_tuc(arr):
    if not arr:
        return []
    n = len(arr)
    dp = [1] * n       
    trace = [-1] * n  
    for i in range(1, n):
        for j in range(i):
            
            if arr[i] >= arr[j] and dp[j] + 1 >= dp[i]:
                dp[i] = dp[j] + 1
                trace[i] = j  
    max_len = max(dp)
    end_index = dp.index(max_len)
    ket_qua = []
    while end_index != -1:
        ket_qua.append(arr[end_index])
        end_index = trace[end_index]  
    ket_qua.reverse()
    return ket_qua
arr=list(map(int,input().split()))
print(*day_con_tang_khong_lien_tuc(arr))
```
