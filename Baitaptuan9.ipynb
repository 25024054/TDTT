Viết chương trình in ra hoán vị ở vị trí tiếp theo (danh sách các hoán vị được sắp xếp theo thứ tự từ điển) của hoán vị đã cho.Đầu vào từ bàn phím gồm 2 dòng:- Dòng đầu tiên chứa số nguyên n - Dòng tiếp theo chứa n số nguyên là một hoán vị của dãy 1,2,...,n (n≤105).Đầu ra: In ra màn hình hoán vị tiếp theo của hoán vị đã cho

```python
def hoan_vi(mang):
    if len(mang) <= 1:
        return [mang]
    ket_qua = []   
    for i in range(len(mang)):
        so_hien_tai = mang[i]
        cac_so_con_lai = mang[:i] + mang[i+1:]
        for p in hoan_vi(cac_so_con_lai):
            ket_qua.append([so_hien_tai] + p)        
    return ket_qua
n=int(input())
mang=list(range(1,n+1))
print(hoan_vi(mang))
```

### W9A10
[Tìm bán kính lò sưởi nhỏ nhất]
Mùa đông đang đến! Bạn được cung cấp vị trí của các ngôi nhà và các lò sưởi trên một
đường thẳng.
Nhiệm vụ của bạn là tìm bán kính r nhỏ nhất cho các lò sưởi sao cho mọi ngôi nhà đều
nằm trong phạm vi sưởi ấm của ít nhất một lò sưởi.
Tất cả các lò sưởi đều có cùng một bán kính r.
Một ngôi nhà tại vị trí h được sưởi ấm bởi một lò sưởi tại vị trí t nếu |h - t| <= r.
Viết hàm `def findRadius(houses, heaters):`
Hàm nhận đầu vào là hai danh sách (list) các số nguyên:
- `houses`: chứa vị trí của các ngôi nhà.
- `heaters`: chứa vị trí của các lò sưởi.
Hàm trả về một số nguyên là bán kính r nhỏ nhất cần thiết.
Lưu ý:
- Số lượng nhà và lò sưởi là một số không âm và không vượt quá 25000.
- Vị trí của các ngôi nhà và lò sưởi là một số không âm và không vượt quá 10^9. 

```python
def findRadius(houses,heaters):
    max=0
    for i in houses:
        for j in heaters:
            if abs(i-j)>max:
                max=abs(i-j)
    return max
```

```python
def power(base, exp):
    if exp == 0:
        return 1     
    half = power(base, exp // 2)
    if exp % 2 == 0:
        return half * half
    else:
        return base * half * half
```

### W9A14 VIế́t hàm tìm số xuất hiện nhiều hơn x lần trong một dãy số (x là số nguyên dương nhập từ bàn phím). Nếu không có số nào thì in ra "INVALID" 

```python
def search(x,lst):
    a=[]
    for i in lst:
        if lst.count(i)>x and i not in a:
            a.append(i)
    if len(a)==0:
        return "INVALID"
    else:
        return a
x=int(input())
lst=list(map(int,input().split()))
print(search(x,lst))
```
