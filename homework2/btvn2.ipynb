### Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.

```python
a=float(input())
b=float(input())
c=float(input())
p=(a+b+c)/2
import math
if a+c>b and a+b>c and b+c>a:
 if a==b==c :
  print("Day la tam giac deu")
  print("Dien tich :",(a**2)*math.sqrt(3)/4)
  print("Chu vi:",a+b+c)
 elif a==b!=c or a==c!=b or b==c!=a:
  print("Day la tam giac can")
  print("Dien tich :",math.sqrt(p*(p-a)*(p-b)*(p-c)))
  print("Chu vi:",a+b+c)
 elif a**2+c**2==b**2 or a**2+b**2==c**2 or b**2+c**2==a**2:
  print("Day la tam giac vuong")
  print("Dien tich :",math.sqrt(p*(p-a)*(p-b)*(p-c)))
  print("Chu vi:",a+b+c)
 elif a!=b!=c:
  print("Day la tam giac thuong")
  print("Dien tich :",math.sqrt(p*(p-a)*(p-b)*(p-c)))
  print("Chu vi:",a+b+c)
else:
 print("Day khong phai tam giac")
   
```

### Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
x=float(input("Nhap toa do x cua tam"))
y=float(input("Nhap toa do y cua tam"))
r=float(input("Nhap ban kinh cua tam"))
x1=float(input("Nhap toa do x cua diem A"))
x2=float(input("Nhap toa do y cua diem A"))
import math
if math.sqrt((x-x1)**2+(y-x2)**2) < r:
    print("Diem A thuoc duong tron")
else:
    print("Diem A khong thuoc duong tron")

```
