### W4A8 Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome

```python
n=int(input())
a=0
while True:
 b=str(n)
 c=b[::-1]
 if b==c:
  break
 else:
  a=a+1
  n=n+int(c)
print(a,n)
```
