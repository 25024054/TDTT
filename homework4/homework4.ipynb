### W4A18- Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b

```python
a=int(input())
b=int(input())
if a>b:
 for i in range(1,a+1):
  if a%i==0 and b%i==0:
   print(i,end=" ")
elif a<b:
 for i in range(1,b+1):
  if a%i==0 and b%i==0:
   print(i,end=" ")
elif a==b:
 for i in range(1,a+1):
  if a%i==0:
   print(i,end=" ")
```

### W4A29 Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành thường và thường thành hoa. 

```python
n=input()
a=""
for i in n: 
 if 65<=ord(i)<=90:
  i=chr(ord(i)+32)
  a=a+i
 elif 97<= ord(i) <= 122:
  i=chr(ord(i)-32)
  a=a+i
 else:
  a=a+i
print(a)
```
