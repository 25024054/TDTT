### W5A16 Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên

```python
def amstrong(n):
 x=0
 if n<1:
  return False
 else:
  for i in str(n):
   x=x+int(i)**len(str(n))
  if x==n:
   return True
  else:
   return False
print(amstrong(371))
```

### W5A25 Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), (0,3)], p = (2,1) 


```python
def tich_cheo(A, B, P):
 return (B[0] - A[0]) * (P[1] - A[1]) - (B[1] - A[1]) * (P[0] - A[0])
def point_in_quadrilateral(quad, p):
    A = quad[0]
    B = quad[1]
    C = quad[2]
    D = quad[3]
    c1 = tich_cheo(A, B, p)
    c2 = tich_cheo(B, C, p)
    c3 = tich_cheo(C, D, p)
    c4 = tich_cheo(D, A, p)
    co_so_am = (c1 < 0) or (c2 < 0) or (c3 < 0) or (c4 < 0)
    co_so_duong = (c1 > 0) or (c2 > 0) or (c3 > 0) or (c4 > 0)
    if co_so_am and co_so_duong:
     return False # P nằm ngoài
    else:
     return True # P nằm trong hoặc trên biên
san_hinh_vuong = [(0,0), (4,0), (4,4), (0,4)]
p_trong = (2, 2)
ket_qua_1 = point_in_quadrilateral(san_hinh_vuong, p_trong)
print( ket_qua_1)
```

### W5A34 Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
 x=0
 if n<=0:
  return 0
 else:
  for i in range(1,n+1):
   if len(set(str(i)))==len(str(i)):
    x=x+i
 return x
print(sum_1_to_n(12))
```
