### W6A16 Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải 

```python
n, m = map(int, input().split())
a = []
for i in range(n):
    row = list(map(int, input().split()))  #nhập hệ số ma trận
    a.append(row)
for i in range(n):
    for j in range(m):
        print(f"{a[i][j]:>4}", end="")
    print()
```

### W6A20 Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1

```python
def timkiem(n,k):
 for i in range(len(n)):
  if n[i]==k:
   return i
 else:
  return -1
n=list(map(int,input().split()))
k=int(input())
print(timkiem(n,k))
```

### W6A21 Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
s=input()
a=s.split()
x={}
for i in range(len(a)):
 key=a[i]
 if key not in x:
  x[key]=i
 else:
  x[key].append(i)
for key in x:
    x[key] = tuple(x[key])
print(x)

```

### W6A24 Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def day(n):
 a=[]
 b=[]
 for i in n:
  if i==0:
   a.append(i)
  else:
   b.append(i)
 return a+b
n=list(map(int,input().split()))
print(day(n))
  
```

### W6A25 Cho dãy chỉ gồm N giá trị từ 0 - 9 . Kiểm tra xem giá trị nào không có trong mảng

### W6A26 Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối 

```python
def tinhtong(n):
 a=0
 for i in range(1,len(n)-1):
  if n[i-1]<n[i] and n[i]>n[i+1]:
   a=a+n[i]
 return a
n=list(map(int,input().split()))
print(tinhtong(n))

  
```

### W6A27 Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). nếu không có in ra -1 

```python
def timkiem(n,x):
 a=[]
 for i in range(len(n)):
  if n[i]<x:
   a.append(i)
 if len(a)==0:
  return [-1]
 else:
  return a
n=list(map(int,input().split()))
x=int(input())
print(*timkiem(n,x))
 
```

### W6A28 Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm)

```python
def kiemtra(n):
 a=0
 b=0
 for i in range(len(n)-1):
  if n[i]<n[i+1]:
   a=a+1
  elif n[i]>n[i+1]:
   b=b+1
 if a!=0 and b==0:
  return "daytang"
 elif a==0 and b!=0:
  return "daygiam"
 else:
  return "vodanh"
n=list(map(int,input().split()))
print(kiemtra(n))
```
