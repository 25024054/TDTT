### W8A16 Yêu cầu Xây dựng lớp Shape với phương thức area(). Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area(). Lấy π = 3.14. Input Dòng 1: loại hình (RECTANGLE hoặc CIRCLE) Dòng 2: tham số tương ứng. Output In ra diện tích với 2 chữ số thập phân. 

```python
class Shape:
    pass
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r
shape_type = input().strip()     
if shape_type == "RECTANGLE":
    w, h = map(float, input().split()) 
    rect = Rectangle(w, h)
    print(f"{rect.area():.2f}")
elif shape_type == "CIRCLE":
    r = float(input())
    circle = Circle(r)
    print(f"{circle.area():.2f}")
```

Output 
Số dư sau khi cộng lãi, in 2 chữ số thập phân

### W8A22 Yêu cầu Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận: 
[a b] 
[c d] 
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc. 
Input 
Một dòng chứa 4 số nguyên a b c d. 
Output 
In ra trace và determinant trên cùng một dòng. 

```python
class Matrix2x2():
    def __init__(self,a,b,c,d):
        self.a=a
        self.b=b
        self.c=c
        self.d=d
    def trace(self):
        return self.a+self.d
    def determinant(self):
        return self.a*self.d-self.b*self.c
a,b,c,d=map(int,input().split())
M1=Matrix2x2(a,b,c,d)
print(f"{M1.trace()} {M1.determinant()}")
```

Input 
Một dòng chuỗi msg. 
Output 
Kết quả của phương thức log().

```python
class BaseLogger():
    def __init__(self,msg):
        self.msg=msg
    def log(self):
        return self.msg
class ErrorLogger(BaseLogger):
    def __init__(self,msg):
        self.msg=msg
    def log(self):
        return f"ERROR: {self.msg}"
msg=input()
B1=ErrorLogger(msg)
print(B1.log())
```

### W8A27 Yêu cầu Xây dựng lớp Product(name, price). Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức final_price() để trả về giá sau giảm. Công thức: price * (100 - percent) / 100  
Input 
name price percent  
Output 
In ra giá cuối cùng với 2 chữ số thập phân. 

```python
class Product():
    def __init__(self,name,price):
        self.name=name
        self.price=float(price)
    def final_price(self):
        return self.price
class DiscountProduct(Product):
    def __init__(self,name,price,percent):
        super().__init__(name,price)
        self.percent=float(percent)
    def final_price(self):
        return self.price*(100-self.percent)/100
name,price,percent=map(str,input().split())
D1=DiscountProduct(name,price,percent)
print(f"{D1.final_price():.2f}")
    
```

### W8A28 Yêu cầu Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), value(). Sau đó thực hiện dãy thao tác trên đối tượng.  
Input 
Dòng 1: n 
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách. 
Output 
Giá trị cuối cùng của count. 

```python
class Counter():
    def __init__(self):
        self.count=0
    def inc(self):
        self.count+=1
    def dec(self):
         self.count-=1
    def value(self):
        return self.count
n=int(input())
s=input().strip()
lst=s.split()
c=Counter()
for i in lst:
    if i =="INC":
        c.inc()
    elif i=="DEC":
        c.dec()
print(c.value())
```

### W8A29 Yêu cầu : Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây. 
Input 
h m s 
Output 
Tổng số giây

```python
class Time():
    def __init__(self,h,m,s):
        self.h=h
        self.m=m
        self.s=s
    def to_seconds(self):
        return self.h*3600+self.m*60+self.s
h,m,s=map(int,input().split())
Time1=Time(h,m,s)
print(Time1.to_seconds())
```
