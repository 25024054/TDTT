### W9A16 Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển 

```python
def generate_binary_strings(n, m, current_string):
    if n==0 and m==0:
        print(current_string)
        return
    if n>0:
        generate_binary_strings(n-1, m, current_string+"0")
    if m>0:
        generate_binary_strings(n, m-1, current_string+"1")
try:
    n, m = map(int, input().split())
    generate_binary_strings((n, m), "")
except ValueError:
    pass
# khong cần print(generate.... vì trên ta đã print roi , cỉ cần gọi hàm nx
#try,except chỉ phòng TH nhập abc chứ k nhập số thì sẽ pass(bỏ qua)chứ k lỗi đỏ
```

```python
def find_LIS_sequence(arr):
    n = len(arr)
    if n == 0:
        return []
    dp = [1] * n
    trace = [-1] * n  
    max_len = 1
    last_index = 0   
    for i in range(1, n):
        for j in range(i):
            if arr[j] < arr[i] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                trace[i] = j  
        if dp[i] > max_len:
            max_len = dp[i]
            last_index = i
    lis = []
    curr = last_index
    while curr != -1:
        lis.append(arr[curr])  
        curr = trace[curr]     
    return lis[::-1]  
try:
    arr = list(map(int, input().split()))
    result_sequence = find_LIS_sequence(arr)
    print("Dãy con tăng nghiêm ngặt dài nhất là:", " ".join(map(str, result_sequence)))
except ValueError:
    print("Vui lòng nhập định dạng số hợp lệ!")
```

### W9A25 Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem nó có phải là mật khẩu mạnh không. 

```python
n=input().strip()
a=0
b=0
c=0
d=0
if len(n)<8:
    print("False")
else:
 for i in n:
    if i.isdigit():
        a+=1
    elif 65<=ord(i)<=90:
       b+=1
    elif 97<=ord(i)<=122:
       c+=1
    elif i in ["!","@","#","$"]:
     d=d+1
 if a!=0 and b!=0 and c!=0 and d!=0:
    print("True")
 else:
    print("False")
```
