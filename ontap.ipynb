### 1 Yêu cầu. Cước phí Taxi Thông Minh. Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả, kiểu số nguyên, đơn vị VND). Các quy tắc cần cài đặt: 1. Giá mở cửa (0–0.8 km): - Xe 4 chỗ: 11,000 VND - Xe 7 chỗ: 12,000 VND 2. Giá theo km sau 0.8 km: - Xe 4 chỗ: từ 0.8→20 km nh 13,000 VND/km; trên 20 km tính 10,500 VND/km - Xe 7 chỗ: từ 0.8→20 km nh 14,000 VND/km; trên 20 km tính 12,000 VND/km 3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND. 4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường + chờ). 5. Điều kiện đặc biệt: - Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền. - Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 

```python
def  smart_taxi_fare(car_type, start_hours, km, wait_minutes):
 total=0
 if car_type=="4cho":
  if wait_minutes>5:
   if (wait_minutes-5)%5==0:
    total=((wait_minutes-5)//5)*2000
   elif (wait_minutes-5)%5!=0:
    total=(((wait_minutes-5)//5)+1)*2000
  if km<=0.8:
   total+=11000
  if 0.8<km<=20:
   total+=11000+(km-0.8)*13000
  if km>20:
   total+=11000+(20-0.8)*13000+(km-20)*10500
  if start_hours in [22,23,0,1,2,3,4,5]:
   total=total*1.2
  if km>30 and start_hours not in [22,23,0,1,2,3,4,5]:
   total=total*0.9
  if total <50000 and km>5:
   total=50000
  return int(total)
 if car_type=="7cho":
  if wait_minutes>5:
   if (wait_minutes-5)%5==0:
    total=((wait_minutes-5)//5)*2000
   elif (wait_minutes-5)%5!=0:
    total=(((wait_minutes-5)//5)+1)*2000
  if km<=0.8:
   total+=12000
  if 0.8<km<=20:
   total+=12000+(km-0.8)*14000
  if km>20:
   total+=12000+(20-0.8)*14000+(km-20)*12000
  if start_hours in [22,23,0,1,2,3,4,5]:
   total=total*1.2
  if km>30 and start_hours not in [22,23,0,1,2,3,4,5]:
   total=total*0.9
  if total <50000 and km>5:
   total=50000
  return int(total)
```

```python
def tich_cheo(A, B, P):
 return (B[0] - A[0]) * (P[1] - A[1]) - (B[1] - A[1]) * (P[0] - A[0])
def point_in_quadrilateral(quad, p):
    A = quad[0]
    B = quad[1]
    C = quad[2]
    D = quad[3]
    c1 = tich_cheo(A, B, p)
    c2 = tich_cheo(B, C, p)
    c3 = tich_cheo(C, D, p)
    c4 = tich_cheo(D, A, p)
    co_so_am = (c1 < 0) or (c2 < 0) or (c3 < 0) or (c4 < 0)
    co_so_duong = (c1 > 0) or (c2 > 0) or (c3 > 0) or (c4 > 0)
    if co_so_am and co_so_duong:
     return False # P nằm ngoài
    else:
     return True # P nằm trong hoặc trên biên
san_hinh_vuong = [(0,0), (4,0), (4,4), (0,4)]
p_trong = (2, 2)
ket_qua_1 = point_in_quadrilateral(san_hinh_vuong, p_trong)
print( ket_qua_1)
```

```python
def detect_number_type(s):
 if len(s)==0:
  return "neither"
 else:
  if s[0]=="+" or s[0]=="-":
   s=s[1:]    #lấy hết cả chuỗi trừ kí tự dầu
  if s.isdigit(): #hàm isdigit() trả về True nếu chuỗi chỉ chứa chữ số (0-9)
   return "int"
  parts=s.split(".")
  if len(parts)==2:
   if parts[0].isdigit() and parts[1].isdigit():
    return "float"
  return "neither"
print(detect_number_type("900.abcđ"))
```

```python
def compute(a, b, op):
 if op=="+":
  return a+b
 elif op=="-":
  return a-b
 elif op=="*":
  return a+b
 elif op=="/":
  if b!=0:
   return a+b
  elif b==0:
   raise ValueError
 else:
  raise ValueError
```

### 35 Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao cho dividend = divisorq + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %). - Đầu vào: số nguyên dividend, divisor (divisor khác 0). - Đầu ra: bộ hai số nguyên (q, r). - Ví dụ: divide_with_remainder(5, 3) ->(1,2)

```python
def  divide_with_remainder(dividend, divisor):
 if divisor==0:
  raise ValueError
 else:
  r=dividend%divisor
  q=dividend//divisor
 return (q,r)
print( divide_with_remainder(5,3))
```

### 43.Viết hàm tính diện tích tam giác có ba cạnh a, b, c (dùng công thức Heron). Phát sinh ValueError nếu ba cạnh không thể tạo thành tam giác. 

```python
def dientich(a,b,c):
 if a+b>c and b+c>a and a+c>b:
  p=(a+b+c)/2
  return (p*(p-a)*(p-b)*(p-c))**0.5
 else:
  return ValueError
print(dientich(6,8,10))
```
